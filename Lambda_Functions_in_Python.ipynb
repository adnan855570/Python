# Difference between Normal and Lambda Functions
## 1. Lambda has no return value
## 2. they are one liner functions
## 3. not used for code reusability
## 4. they have no names

```python
b = lambda x:x[0]=='a'
b('apple')
```

```python
b('banana')
```

```python
c = lambda x: "Even" if x%2 == 0 else "Odd"
c(5)
```

```python
c(2)
```

# Higher Order Functions

```python
def return_sum(func,L):
  result=0
  for i in L:
    if func(i):
      result = result + i

  return result

L = [11,14,21,23,56,78,0,2]
x = return_sum(lambda x:x%2==0,L)
print(x)
y = return_sum(lambda x:x%2!=0,L)
print(y)
```

# Map function

```python
a = [1,23,4,6,7,8,9]

list(map(lambda x:x*2,a))
```

```python
list(map(lambda x:x%2==0,a))
```

# Filter Function

```python
fruits = ['apple', 'banana', 'cherry', 'date', 'elderberry']

list(filter(lambda x: x.startswith('a'), fruits))
```

# List Comprehension

```python
c = [1,2,3,4,5,6,7,8,9,0]

c = [x*2 for x in c]
c
```

```python
c1 = [i**2 for i in range(10)]
c1
```

```python
c2 = [i**2 for i in range(10) if i%2!=0]
c2
```
