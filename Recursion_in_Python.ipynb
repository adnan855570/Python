```python
def mul(a,b):
  if b == 1:
    return a
  else:
    return a + mul(a,b-1)

print(mul(4,5))
```

```python
def factorial(n):
    # Base case
    if n == 0 or n == 1:
        return 1

    # Recursive case
    return n * factorial(n - 1)


print(factorial(5))
```
